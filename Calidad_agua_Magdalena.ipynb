## Descripción del conjunto de datos

Este conjunto de datos fue descargado de REDCAM mediante SIAM Explorer
y corresponde a registros de calidad del agua del departamento del
Magdalena, Colombia.

Contiene 45.094 registros y 21 columnas. Cada fila presenta el resultado
de una variable medida en una muestra, junto con información como la
estación, el lugar, la fecha, la temporada, el método y las unidades.

Incluye turbidez (NTU), temperatura (°C), pH, oxígeno disuelto (mg/L)
y sólidos disueltos totales (mg/L). El pH es adimensional.

Se identificaron también registros de temperatura con unidades de %.
Esta combinación requiere verificar los datos originales antes de
realizar análisis; en esta actividad se conserva tal como fue descargada.

El objetivo del notebook es cargar un archivo CSV y explorar sus
primeras filas.
